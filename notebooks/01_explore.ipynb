### Data Integration and Cleaning Decisions

- The analysis will focus exclusively on crime events that occurred in 2025.
- The crime categories included in the analysis are 
    - `ASSAULT`
    - `AUTHOTHEFT`
    - `BREAKENTER`
    - `HOMICIDE`
    - `ROBBERY`
    - `SHOOTING`
    - `THEFTFROMMV`
    - `THEFTOVER`
- Only the variables required for analysis will be retained, including 
    - `_id`
    - `AREA_NAME`
    - `HOOD_ID`
    - `POPULATION`
- In the homicide and shooting datasets, `HOOD_158` will be converted from a string to an integer and renamed to `HOOD_ID` to ensure consistency across datasets.
- Homicide and shooting data are recorded at the incident level. These records will therefore be aggregated by `HOOD_ID` to calculate the number of occurrences in each neighbourhood.
- After the homicide and shooting datasets are merged with the neighbourhood crime data, remaining missing values for these two crime categories will be replaced with 0, indicating that no recorded homicide or shooting occurred in that neighbourhood during 2025.
- Following Statistics Canada’s standard convention, homicide and shooting rates will be calculated as the number of incidents per 100,000 residents.
