## Data Preprocessing

### Drop Irrelevant Columns

- The analysis will focus exclusively on crime events that occurred in 2025.
- The crime categories included in the analysis are 
    - `ASSAULT`
    - `AUTHOTHEFT`
    - `BREAKENTER`
    - `HOMICIDE`
    - `ROBBERY`
    - `SHOOTING`
    - `THEFTFROMMV`
    - `THEFTOVER`
- Only the variables required for analysis will be retained, including 
    - `_id`
    - `AREA_NAME`
    - `HOOD_ID`
    - `POPULATION`

```python
# keep only columns of interest
crime_clean = crime[[
    "_id",
    "AREA_NAME",
    "HOOD_ID",
    "ASSAULT_2025",
    "ASSAULT_RATE_2025",
    "AUTOTHEFT_2025",
    "AUTOTHEFT_RATE_2025",
    "BREAKENTER_2025",
    "BREAKENTER_RATE_2025",
    "HOMICIDE_2025",
    "HOMICIDE_RATE_2025",
    "ROBBERY_2025",
    "ROBBERY_RATE_2025",
    "SHOOTING_2025",
    "SHOOTING_RATE_2025",
    "THEFTFROMMV_2025",
    "THEFTFROMMV_RATE_2025",
    "THEFTOVER_2025",
    "THEFTOVER_RATE_2025",
    "POPULATION_2025"
    ]].copy()
```

### Handling Missing Values

Columns with missing data will be cross-referenced against supplementary Toronto Police datasets to reconstruct the missing values where possible.



### Supplementary Homicide Data Set

#### Loading and Initial Inspection

```python
homicide = pd.read_csv("../data/raw/Homicides - 4326.csv")
```

#### Basic Data Exploration (filtered to 2025 records only)

```python
## Show rows and columns count 
print(f"Rows count: {homicide[homicide["OCC_YEAR"] == 2025].shape[0]}\nColumns count: {homicide[homicide["OCC_YEAR"] == 2025].shape[1]}")
```

```python
homicide[homicide["OCC_YEAR"] == 2025].head()
```

```python
homicide[homicide["OCC_YEAR"] == 2025].tail()
```

#### Check Data Types

```python
homicide.dtypes
```

#### Column Alignment

The column that corresponds to the neighborhood identifier
- `HOOD_158` from **homicide** and `HOOD_ID` from **crime**

```python
# keep only homicide records from 2025
homicide_2025 = homicide[homicide["OCC_YEAR"] == 2025].copy()

# Convert neighbourhood ID from string to integer
homicide_2025["HOOD_158"] = homicide_2025["HOOD_158"].astype(int)

# standardize neighbourhoud ID column
homicide_2025 = homicide_2025.rename(columns={
    "HOOD_158": "HOOD_ID"
})
```

#### Check for Missing Data 

```python
homicide_2025.isna().sum()
```

#### Check for Duplicate Rows 

```python
homicide_2025.duplicated().sum()
```

#### Comparison

Compare 2025 homicide incidents with the neighbourhood crime dataset to confirm the values are consistent and identify any discrepancies.

```python
# Group the number of homicide incidents in each neighbourhood
homicide_count = (
    homicide_2025[["HOOD_ID"]]
    .groupby("HOOD_ID")
    .size()
    .reset_index(name="HOMICIDE_2025_NEW")
)

# Create new dataframe with the updated homicide counts
compare_01 = crime_clean.merge(
    homicide_count,
    on="HOOD_ID",
    how="left"
)

# No matching homicide record = 0 homicides
compare_01["HOMICIDE_2025_NEW"] = (
    compare_01["HOMICIDE_2025_NEW"]
    .fillna(0)
    .astype(int)
)

# Compare old and newly calculated homicide counts
print(compare_01[["HOOD_ID", "HOMICIDE_2025","HOMICIDE_2025_NEW"]])
```

- Comparison with the *homicides* dataset confirmed that all missing `HOMICIDE_2025` values in the *neighbourhood crime* dataset correspond to neighbourhoods with **zero recorded homicides**.
- These missing values can therefore be safely replaced with 0, rather than treated as unknown or incomplete observations.

```python
# fill NaN values in HOMICIDE_2025 and HOMICIDE_RATE_2025 columns with 0
crime_clean[["HOMICIDE_2025", "HOMICIDE_RATE_2025"]] = (
    crime_clean[["HOMICIDE_2025", "HOMICIDE_RATE_2025"]]
    .fillna(0)
)
```

### Supplementary Shooting Data Set

#### Loading and Initial Inspection

```python
shooting = pd.read_csv("../data/raw/shootings-firearm-discharges - 4326.csv")
```

#### Basic Data Exploration (filtered to 2025 records only)

```python
## Show rows and columns count
print(f"Rows count: {shooting[shooting["OCC_YEAR"]==2025].shape[0]}\nColumns count: {shooting[shooting["OCC_YEAR"]==2025].shape[1]}")
```

```python
shooting[shooting["OCC_YEAR"]==2025].head()
```

```python
shooting[shooting["OCC_YEAR"]==2025].tail()
```

#### Check Data Types

```python
shooting.dtypes
```

### Column Alignment 

The column that corresponds to the neighborhood identifier
- `HOOD_158` from **shooting** and `HOOD_ID` from **crime**

```python
# keep only shooting & firearm records from 2025
shooting_2025 = shooting[
    (shooting["OCC_YEAR"] == 2025) 
].copy()

# Remove records without a valid neighbourhood ID
shooting_2025 = shooting_2025[
    shooting_2025["HOOD_158"] != "NSA"
].copy()

# Convert neighbourhood ID from string to integer
shooting_2025["HOOD_158"] = shooting_2025["HOOD_158"].astype(int)

# standardize neighbourhoud ID column
shooting_2025 = shooting_2025.rename(columns={
    "HOOD_158": "HOOD_ID"
})
```

#### Check for Missing Data

```python
shooting_2025.isna().sum()
```

Missing values in the `DEATH` and `INJURIES` columns are expected because a shooting event may result in injuries, deaths, or both. A missing value in one of these columns does not necessarily indicate incomplete data; it may simply mean that outcome did not occur for that incident.

#### Check for Duplicate Rows

```python
print("Shooting duplicates:", shooting_2025.duplicated().sum())
```

#### Comparison

Compare 2025 shooting & firearm incidents with the neighbourhood crime dataset to confirm the values are consistent and identify any discrepancies.

```python
# Group the number of shooting incidents in each neighbourhood
shooting_count = (
    shooting_2025[["HOOD_ID"]]
    .groupby("HOOD_ID")
    .size()
    .reset_index(name="SHOOTING_2025_NEW")
)

# Create new dataframe with the updated shooting counts
compare_02 = crime_clean.merge(
    shooting_count,
    on="HOOD_ID",
    how="left"
)

# No matching shooting record = 0 shootings
compare_02["SHOOTING_2025_NEW"] = (
    compare_02["SHOOTING_2025_NEW"]
    .fillna(0)
    .astype(int)
)

# Compare old and newly calculated shooting counts
print(compare_02[["HOOD_ID", "SHOOTING_2025","SHOOTING_2025_NEW"]])
```

- Comparison with the *shooting* dataset confirmed that all missing `SHOOTING_2025` values in the *neighbourhood crime* dataset correspond to neighbourhoods with **zero recorded shooting & firearm charges**.
- These missing values can therefore be safely replaced with 0, rather than treated as unknown or incomplete observations.

```python
# fill NaN values in SHOOTING_2025 and SHOOTING_RATE_2025 columns with 0
crime_clean[["SHOOTING_2025", "SHOOTING_RATE_2025"]] = (
    crime_clean[["SHOOTING_2025", "SHOOTING_RATE_2025"]]
    .fillna(0)
)
```

```python
# verify that all missing values were handled
assert crime_clean.isna().sum().sum() == 0
```

## Exploratory Data Analysis

```python
all_rates = [
    "ASSAULT_RATE_2025",
    "AUTOTHEFT_RATE_2025",
    "BREAKENTER_RATE_2025",
    "HOMICIDE_RATE_2025",
    "ROBBERY_RATE_2025",
    "SHOOTING_RATE_2025",
    "THEFTOVER_RATE_2025",
    "THEFTFROMMV_RATE_2025",
]

crime_clean["TOTAL_CRIME_RATE_2025"] = (
    crime_clean[all_rates].sum(axis=1)
)

# creating violent crime category
violent_crime_rates = [
    "ASSAULT_RATE_2025",
    "HOMICIDE_RATE_2025",
    "ROBBERY_RATE_2025",
    "SHOOTING_RATE_2025",
]

crime_clean["VIOLENT_RATE_2025"] = (
    crime_clean[violent_crime_rates].sum(axis=1)
)

# creating property crime category
property_crime_rates = [
    "AUTOTHEFT_RATE_2025",
    "BREAKENTER_RATE_2025",
    "THEFTOVER_RATE_2025",
    "THEFTFROMMV_RATE_2025",
]


crime_clean["PROPERTY_RATE_2025"] = (
    crime_clean[property_crime_rates].sum(axis=1)
)
```

## Data Visualization

### Exploring Total Crime Rate

```python
all_rates = [
    "ASSAULT_RATE_2025",
    "AUTOTHEFT_RATE_2025",
    "BREAKENTER_RATE_2025",
    "HOMICIDE_RATE_2025",
    "ROBBERY_RATE_2025",
    "SHOOTING_RATE_2025",
    "THEFTOVER_RATE_2025",
    "THEFTFROMMV_RATE_2025",
]

crime_clean["TOTAL_CRIME_RATE_2025"] = (
    crime_clean[all_rates].sum(axis=1)
)

crime_rate_ranking = crime_clean[
    ["HOOD_ID", "AREA_NAME", "TOTAL_CRIME_RATE_2025"]
].sort_values(
    "TOTAL_CRIME_RATE_2025",
    ascending=False
).head(10)

display(crime_rate_ranking)
```

### Exploring Top/Bottom Violent Crime Rates

```python
violent_crime_rates = [
    "ASSAULT_RATE_2025",
    "HOMICIDE_RATE_2025",
    "ROBBERY_RATE_2025",
    "SHOOTING_RATE_2025",
]

crime_clean["VIOLENT_RATE_2025"] = (
    crime_clean[violent_crime_rates].sum(axis=1)
)

top_violent = crime_clean[
    ["HOOD_ID", "AREA_NAME", "VIOLENT_RATE_2025"]
].sort_values(
    "VIOLENT_RATE_2025",
    ascending=False
).head(10)

# Plot
plt.figure(figsize=(10, 6))
plt.bar(top_violent["AREA_NAME"], top_violent["VIOLENT_RATE_2025"])

plt.xlabel("Neighbourhood")
plt.xticks(fontsize=10, rotation=45, ha='right')
plt.ylabel("Violent Crime Rate per 100,000")
plt.title("Top 10 Neighbourhoods by Violent Crime Rate, 2025")

plt.tight_layout()
plt.show()
```

```python
bottom_violent = crime_clean[
    ["HOOD_ID", "AREA_NAME", "VIOLENT_RATE_2025"]
].sort_values(
    "VIOLENT_RATE_2025",
    ascending=True
).head(10)

# Plot
plt.figure(figsize=(10, 6))
plt.bar(bottom_violent["AREA_NAME"], bottom_violent["VIOLENT_RATE_2025"])
plt.ylim(0, 4000)


plt.xlabel("Neighbourhood")
plt.xticks(fontsize=10, rotation=45, ha='right')
plt.ylabel("Violent Crime Rate per 100,000")
plt.title("Bottom 10 Neighbourhoods by Violent Crime Rate, 2025")

plt.tight_layout()
plt.show()
```
