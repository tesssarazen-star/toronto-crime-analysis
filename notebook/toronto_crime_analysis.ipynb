# 1
## Introduction

### Project Description

**Goal:**
- Analyze 2025 crime patterns across Toronto neighbourhoods
- Identify neighborhoods with the highest levels of violent crime, property crime, and overall crime
- Examine how each crime category is distributed to understand which specific offences drive high crime rates across Toronto neighborhoods

**Questions to be Answered:**
- Which neighbourhoods have the highest violent crime rates in 2025?
- Which neighbourhoods have the highest property crime rates in 2025?
- Which neighbourhoods rank highest when violent and property crime are combined? 
- What proportion of overall crime is made up of violent crime versus property crime?
- Within violent crime, which offences contribute most to high rates?
- Within property crime, which offences contribute most to high rates?
- Do neighbourhoods with high violent crime rates also tend to have high property crime rates?

**Methodology and Scope:**

- The analysis is limited to Toronto crime events occurring in 2025. 
- Crime is grouped into two main categories:
    - Violent crime: assault, robbery, homicide, and shooting.
    - Property crime: auto theft, break and enter, theft from motor vehicle, and theft over $5000.
- Neighbourhoods are ranked by violent, property, and overall crime rates.
- Separate homicide and shooting datasets are used to validate missing values in the main crime dataset before cleaning.
- Key visualization techniques include:
    - histograms
    - boxplots
    - bar charts
    - scatter plots
- The analysis focuses on patterns, composition, and relationships, not causal explanations.

### Data Description

**Content:**

- The main dataset is a CSV file containing 158 neighbourhood level data points, with one row representing each Toronto neighbourhood included in the analysis.
- The dataset contains 2025 crime counts, crime rates, neighbourhood identifiers, area names, and population estimates.
- Crime is analyzed across eight offence types and later grouped into violent crime and property crime categories.

**Acknowledgements:**

The crime datasets used in this project are provided by the **Toronto Police Service Public Safety Data Portal**.
The analysis uses publicly available neighbourhood crime, homicide, and shooting data published by the Toronto Police Service. The original datasets can be accessed through the **Toronto Police Service Public Safety Data Portal / Open Data** resources.

## Data Preprocessing

### Drop Irrelevant Columns

- The analysis will focus exclusively on crime events that occurred in 2025.
- The crime categories included in the analysis are 
    - `ASSAULT`
    - `AUTHOTHEFT`
    - `BREAKENTER`
    - `HOMICIDE`

```python
# keep only columns of interest
crime_clean = crime[[
    "_id",
    "AREA_NAME",
    "HOOD_ID",
    "ASSAULT_2025",
    "ASSAULT_RATE_2025",
    "AUTOTHEFT_2025",
    "AUTOTHEFT_RATE_2025",
    "BREAKENTER_2025",
    "BREAKENTER_RATE_2025",
    "HOMICIDE_2025",
    "HOMICIDE_RATE_2025",
    "ROBBERY_2025",
    "ROBBERY_RATE_2025",
    "SHOOTING_2025",
    "SHOOTING_RATE_2025",
    "THEFTFROMMV_2025",
    "THEFTFROMMV_RATE_2025",
    "THEFTOVER_2025",
    "THEFTOVER_RATE_2025",
    "POPULATION_2025"
    ]].copy()
```

### Handling Missing Values

Columns with missing data will be cross-referenced against supplementary Toronto Police datasets to reconstruct the missing values where possible.

### Supplementary Homicide Data Set

#### Loading and Initial Inspection

```python
homicide = pd.read_csv("../data/raw/Homicides - 4326.csv")
```

#### Basic Data Exploration (filtered to 2025 records only)

```python
# Show rows and columns count 
print(f"Rows count: {homicide[homicide["OCC_YEAR"] == 2025].shape[0]}\nColumns count: {homicide[homicide["OCC_YEAR"] == 2025].shape[1]}")
```

#### Check Data Types

```python
homicide.dtypes
```

#### Column Alignment

The column that corresponds to the neighborhood identifier
- `HOOD_158` from **homicide** and `HOOD_ID` from **crime**

```python
# keep only homicide records from 2025
homicide_2025 = homicide[homicide["OCC_YEAR"] == 2025].copy()

# Convert neighbourhood ID from string to integer
homicide_2025["HOOD_158"] = homicide_2025["HOOD_158"].astype(int)

# standardize neighbourhoud ID column
homicide_2025 = homicide_2025.rename(columns={
    "HOOD_158": "HOOD_ID"
})
```

#### Check for Missing Data 

```python
homicide_2025.isna().sum()
```

#### Check for Duplicate Rows 

```python
homicide_2025.duplicated().sum()
```

#### Comparison

Compare 2025 homicide incidents with the neighbourhood crime dataset to confirm the values are consistent and identify any discrepancies.

```python
# Group the number of homicide incidents in each neighbourhood
homicide_count = (
    homicide_2025[["HOOD_ID"]]
    .groupby("HOOD_ID")
    .size()
    .reset_index(name="HOMICIDE_2025_NEW")
)

# Create new dataframe with the updated homicide counts
compare_01 = crime_clean.merge(
    homicide_count,
    on="HOOD_ID",
    how="left"
)

# No matching homicide record = 0 homicides
compare_01["HOMICIDE_2025_NEW"] = (
    compare_01["HOMICIDE_2025_NEW"]
    .fillna(0)
    .astype(int)
)

# Compare old and newly calculated homicide counts
print(compare_01[["HOOD_ID", "HOMICIDE_2025","HOMICIDE_2025_NEW"]])
```

- Comparison with the *homicides* dataset confirmed that all missing `HOMICIDE_2025` values in the *neighbourhood crime* dataset correspond to neighbourhoods with **zero recorded homicides**.
- These missing values can therefore be safely replaced with 0, rather than treated as unknown or incomplete observations.

```python
# fill NaN values in HOMICIDE_2025 and HOMICIDE_RATE_2025 columns with 0
crime_clean[["HOMICIDE_2025", "HOMICIDE_RATE_2025"]] = (
    crime_clean[["HOMICIDE_2025", "HOMICIDE_RATE_2025"]]
    .fillna(0)
)
```

### Supplementary Shooting Data Set

#### Loading and Initial Inspection

```python
shooting = pd.read_csv("../data/raw/shootings-firearm-discharges - 4326.csv")
```

#### Basic Data Exploration (filtered to 2025 records only)

```python
## Show rows and columns count
print(f"Rows count: {shooting[shooting["OCC_YEAR"]==2025].shape[0]}\nColumns count: {shooting[shooting["OCC_YEAR"]==2025].shape[1]}")
```

#### Check Data Types

```python
shooting.dtypes
```

#### Column Alignment 

The column that corresponds to the neighborhood identifier
- `HOOD_158` from **shooting** and `HOOD_ID` from **crime**

```python
# keep only shooting & firearm records from 2025
shooting_2025 = shooting[
    (shooting["OCC_YEAR"] == 2025) 
].copy()

# Remove records without a valid neighbourhood ID
shooting_2025 = shooting_2025[
    shooting_2025["HOOD_158"] != "NSA"
].copy()

# Convert neighbourhood ID from string to integer
shooting_2025["HOOD_158"] = shooting_2025["HOOD_158"].astype(int)

# standardize neighbourhoud ID column
shooting_2025 = shooting_2025.rename(columns={
    "HOOD_158": "HOOD_ID"
})
```

#### Check for Missing Data

```python
shooting_2025.isna().sum()
```

Missing values in the `DEATH` and `INJURIES` columns are expected because a shooting event may result in injuries, deaths, or both. A missing value in one of these columns does not necessarily indicate incomplete data; it may simply mean that outcome did not occur for that incident.

#### Check for Duplicate Rows

```python
print("Shooting duplicates:", shooting_2025.duplicated().sum())
```

#### Comparison

Compare 2025 shooting & firearm incidents with the neighbourhood crime dataset to confirm the values are consistent and identify any discrepancies.

```python
# Group the number of shooting incidents in each neighbourhood
shooting_count = (
    shooting_2025[["HOOD_ID"]]
    .groupby("HOOD_ID")
    .size()
    .reset_index(name="SHOOTING_2025_NEW")
)

# Create new dataframe with the updated shooting counts
compare_02 = crime_clean.merge(
    shooting_count,
    on="HOOD_ID",
    how="left"
)

# No matching shooting record = 0 shootings
compare_02["SHOOTING_2025_NEW"] = (
    compare_02["SHOOTING_2025_NEW"]
    .fillna(0)
    .astype(int)
)

# Compare old and newly calculated shooting counts
print(compare_02[["HOOD_ID", "SHOOTING_2025","SHOOTING_2025_NEW"]])
```

- Comparison with the *shooting* dataset confirmed that all missing `SHOOTING_2025` values in the *neighbourhood crime* dataset correspond to neighbourhoods with **zero recorded shooting & firearm charges**.
- These missing values can therefore be safely replaced with 0, rather than treated as unknown or incomplete observations.

```python
# fill NaN values in SHOOTING_2025 and SHOOTING_RATE_2025 columns with 0
crime_clean[["SHOOTING_2025", "SHOOTING_RATE_2025"]] = (
    crime_clean[["SHOOTING_2025", "SHOOTING_RATE_2025"]]
    .fillna(0)
)
```

```python
# verify that all missing values were handled
assert crime_clean.isna().sum().sum() == 0
```

### Creating Attributes

```python
# creating violent crime category
violent_crime_rates = [
    "ASSAULT_RATE_2025",
    "HOMICIDE_RATE_2025",
    "ROBBERY_RATE_2025",
    "SHOOTING_RATE_2025",
]

crime_clean["VIOLENT_RATE_2025"] = (
    crime_clean[violent_crime_rates].sum(axis=1)
)

# creating property crime category
property_crime_rates = [
    "AUTOTHEFT_RATE_2025",
    "BREAKENTER_RATE_2025",
    "THEFTOVER_RATE_2025",
    "THEFTFROMMV_RATE_2025",
]


crime_clean["PROPERTY_RATE_2025"] = (
    crime_clean[property_crime_rates].sum(axis=1)
)

# Calculate overall crime rate
crime_clean["TOTAL_CRIME_RATE_2025"] = (
    crime_clean["VIOLENT_RATE_2025"]
    + crime_clean["PROPERTY_RATE_2025"]
)
```

- `VIOLENT_RATE_2025` combines assault, robbery, homicide, and shooting rates.
- `PROPERTY_RATE_2025` combines auto theft, break and enter, theft from motor vehicle, and theft over rates.
- `TOTAL_CRIME_RATE_2025` combines the violent and property crime rates.
- These derived attributes are added directly to the `crime_clean` DataFrame for use in later analysis and visualization.

## Data Visualization

### Top 10 Neighbourhoods by Violent Crime Rate

```python
# Calculate total violent crime rate
crime_clean["VIOLENT_RATE_2025"] = (
    crime_clean[violent_crime_rates].sum(axis=1)
)

# Get top 10 neighbourhoods by violent crime rate
top_violent = crime_clean[
    ["HOOD_ID", "AREA_NAME", "VIOLENT_RATE_2025"]
].sort_values(
    "VIOLENT_RATE_2025",
    ascending=False
).head(10)

# Get violent crime components for those top 10 neighbourhoods
violent_composition = crime_clean[
    crime_clean["HOOD_ID"].isin(top_violent["HOOD_ID"])
][
    [
        "HOOD_ID",
        "AREA_NAME",
        "ASSAULT_RATE_2025",
        "ROBBERY_RATE_2025",
        "HOMICIDE_RATE_2025",
        "SHOOTING_RATE_2025"
    ]
]

# Keep the same ranking order
violent_composition = violent_composition.merge(
    top_violent[["HOOD_ID", "VIOLENT_RATE_2025"]],
    on="HOOD_ID"
).sort_values(
    "VIOLENT_RATE_2025",
    ascending=True
)

# Set neighbourhood names as labels
plot_data = violent_composition.set_index("AREA_NAME")[
    [
        "ASSAULT_RATE_2025",
        "ROBBERY_RATE_2025",
        "HOMICIDE_RATE_2025",
        "SHOOTING_RATE_2025"
    ]
]

# Stacked horizontal bar chart
plot_data.plot(
    kind="barh",
    stacked=True,
    figsize=(10, 7)
)

plt.xlabel("Crime Rate per 100,000")
plt.ylabel("Neighbourhood")
plt.title("Violent Crime Composition of Top 10 Neighbourhoods, 2025")

plt.legend(
    ["Assault", "Robbery", "Homicide", "Shooting"],
    title="Crime Type"
)

plt.tight_layout()
plt.show()
```

### Top 10 Neighbourhoods by Property Crime Rate

```python
# Calculate total property crime rate
crime_clean["PROPERTY_RATE_2025"] = (
    crime_clean[property_crime_rates].sum(axis=1)
)

# Get top 10 neighbourhoods by property crime rate
top_property = crime_clean[
    ["HOOD_ID", "AREA_NAME", "PROPERTY_RATE_2025"]
].sort_values(
    "PROPERTY_RATE_2025",
    ascending=False
).head(10)

# Get property crime components for those top 10 neighbourhoods
property_composition = crime_clean[
    crime_clean["HOOD_ID"].isin(top_property["HOOD_ID"])
][
    [
        "HOOD_ID",
        "AREA_NAME",
        "AUTOTHEFT_RATE_2025",
        "BREAKENTER_RATE_2025",
        "THEFTFROMMV_RATE_2025",
        "THEFTOVER_RATE_2025"
    ]
]

# Keep the same ranking order
property_composition = property_composition.merge(
    top_property[["HOOD_ID", "PROPERTY_RATE_2025"]],
    on="HOOD_ID"
).sort_values(
    "PROPERTY_RATE_2025",
    ascending=True
)

# Set neighbourhood names as labels
plot_data = property_composition.set_index("AREA_NAME")[
    [
        "AUTOTHEFT_RATE_2025",
        "BREAKENTER_RATE_2025",
        "THEFTFROMMV_RATE_2025",
        "THEFTOVER_RATE_2025"
    ]
]

# Stacked horizontal bar chart
plot_data.plot(
    kind="barh",
    stacked=True,
    figsize=(10, 7)
)

plt.xlabel("Crime Rate per 100,000")
plt.ylabel("Neighbourhood")
plt.title("Property Crime Composition of Top 10 Neighbourhoods, 2025")

plt.legend(
    ["Auto Theft", "Break and Enter", "Theft From Motor Vehicle", "Theft Over"],
    title="Crime Type"
)

plt.tight_layout()
plt.show()
```

### Overall Crime Rate Ranking 

```python
# Get top 10 neighbourhoods by overall crime rate
top_overall = crime_clean[
    [
        "HOOD_ID",
        "AREA_NAME",
        "VIOLENT_RATE_2025",
        "PROPERTY_RATE_2025",
        "TOTAL_CRIME_RATE_2025"
    ]
].sort_values(
    "TOTAL_CRIME_RATE_2025",
    ascending=False
).head(10)

# Sort so highest appears at the top of horizontal chart
top_overall = top_overall.sort_values(
    "TOTAL_CRIME_RATE_2025",
    ascending=True
)

# Set neighbourhood names as labels
plot_data = top_overall.set_index("AREA_NAME")[
    ["VIOLENT_RATE_2025", "PROPERTY_RATE_2025"]
]

# Create stacked horizontal bar chart
plot_data.plot(
    kind="barh",
    stacked=True,
    figsize=(10, 7)
)

plt.xlabel("Crime Rate per 100,000")
plt.ylabel("Neighbourhood")
plt.title("Top 10 Neighbourhoods by Overall Crime Rate, 2025")

plt.legend(
    ["Violent Crime", "Property Crime"],
    title="Crime Category"
)

plt.tight_layout()
plt.show()
```

## Conclusion

### Insights

- **Assault** constitutes the dominant component of violent crime rates across nearly all of the highest ranked neighbourhoods.
- **Mimico Queensway** records the highest violent crime rate among the top 10 neighbourhoods
- **Yonge Bay Corridor** also exhibits an elevated violent crime rate with a comparatively larger robbery contribution. 
- Property crime composition varies more substantially across neighbourhoods than violent crime composition, with different areas driven by different combinations of auto theft, break and enter, theft from motor vehicles, and theft over.
- **West Humber Clairville** records the highest property crime rate among the top 10 neighbourhoods, with auto theft representing the largest component of its property crime profile.
- **Kensington Chinatown** exhibits a different property crime structure, with theft from motor vehicles and break and enter accounting for a larger share of the total.
- Overall crime rate rankings indicate that neighbourhoods with similar total crime rates can have materially different underlying crime compositions.
- **Yonge Bay Corridor** and **Mimico Queensway** are primarily characterized by violent crime, whereas **West Humber Clairville** is primarily characterized by property crime.
- **Kensington Chinatown** and **University** display a more balanced composition between violent and property crime.
- Crime category decomposition provides more analytical value than aggregate crime rates alone because it reveals the specific offence types driving neighbourhood level differences.

### Limitations

- The 2025 time scope restricts the analysis to a single cross sectional period and prevents assessment of persistence, seasonality, or longer term crime trends.
- The selected offence set includes only eight crime categories, so `TOTAL_CRIME_RATE_2025` should be interpreted as the total rate for the selected categories rather than a complete measure of all reported crime.
- Residential population serves as the denominator for crime rates, which may distort comparisons in neighbourhoods with high non resident activity from commuters and tourists.
- Geographic location offsetting by Toronto Police may introduce minor neighbourhood classification error, particularly for incidents occurring near neighbourhood boundaries.
- Observed correlations measure linear association only and do not establish causal relationships between population, violent crime, property crime, or individual offence types.
- Extreme neighbourhood rates *may* disproportionately influence summary statistics, correlations, and rankings.
